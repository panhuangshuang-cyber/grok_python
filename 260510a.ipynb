```python
from openpyxl import load_workbook

# 1. 加载（关键：keep_vba=True）
wb = load_workbook('C:/Users/phsjy/Desktop/库存.xlsm', keep_vba=True)

# 2. 清洗+写入数据
ws = wb.active  # 或 wb['Sheet1']
# 示例：清洗 + 写入
for row in ws.iter_rows(min_row=2):  # 跳过标题
    for cell in row:
        if cell.value is None or str(cell.value).strip() == '':  # 简单清洗
            cell.value = None  # 或其他逻辑

# 写入新数据
ws['A10'] = '新数据'
# 或用pandas快速写：
# import pandas as pd
# df = pd.read_excel(...)  # 处理后
# with pd.ExcelWriter('output.xlsm', engine='openpyxl') as writer:
#     df.to_excel(writer, index=False)  # 但需先load再写

# 3. 保存（必须保持原格式）
wb.save('cleaned_file.xlsm')
```
